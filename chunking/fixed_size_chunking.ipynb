```python
with open('data.txt') as file:
    data = file.read()
# print(data)
chunks = []
fixed_size = 20
overlap = 4
for index in range(0,len(data),fixed_size-overlap): #0 20
    chunk = data[index:index+fixed_size] #data[0:0+20] data[0:20] #data[20:20+20]
    # print(chunk,len(chunk))
    chunks.append(chunk)
print(chunks)
```
