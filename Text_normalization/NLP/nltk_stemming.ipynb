```python
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
```

```python
words = [
    "Running", "Runs", "ran", "easily", "fairness",
    "Organization", "organizational", "organize", "organizes", "organizing",
    "activation", "activated", "activating", "activations", "connection",
    "connected", "connecting", "connections", "studies", "studying",'a','the'
]

for word in words:
    print(f'{word} ---> {stemmer.stem(word)}')
```

```python
from nltk.stem import SnowballStemmer
snowball_stemmer = SnowballStemmer('english',ignore_stopwords=False)
```

```python
snowball_words = [
    "running", "runs", "ran", "easily", "fairness",
    "organization", "organizational", "organize", "organizes", "organizing",
    "activation", "activated", "activating", "activations", "connection",
    "connected", "connecting", "connections", "studies", "studying",'a','the','and'
]
```

```python
for word in words:
    print(f"{word} ---> {snowball_stemmer.stem(word)}")
```

```python
from nltk.corpus import stopwords
from nltk.stem import SnowballStemmer
stop_words = stopwords.words("english")
# stop_words
snowball_stemmer = SnowballStemmer("english",ignore_stopwords=True)
snowball_stemmer.stopwords = stop_words
# snowball_stemmer.stopwords
for word in snowball_words:
    print(word,snowball_stemmer.stem(word))
```
