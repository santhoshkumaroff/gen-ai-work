```python
# has a relationship ?

''' 
Has a relationship means one object contains another object (implemented by creating an object of class
inside another class)
'''

class Address:
    def __init__(self):
        self.address='main address..........'

class Student:
    def __init__(self):
        self.name='name....'
        self.age ='age....'
        self.address = Address()

obj  = Student()
print(obj)
print(obj.name)
print(obj.age)
print(obj.address)
```

```python
# what is polymorphism ?

# poly = many,morphism = forms

#  *

# print(10*5)
# print(5*'hello')
# 1)Payment system
# pay()
#runtime polymorphism(method overriding)
class Payment:
    def pay(self,mode):
        print(f'Welcome to {mode} Mode')
class creditcard(Payment):
    def pay(self,mode='credit card'):
        super().pay(mode)
class upi(Payment):
    def pay(self,mode='upi'):
        super().pay(mode) 
class net_banking(Payment):
    def pay(self,mode='net banking'):
        super().pay(mode) 
obj = net_banking()
obj.pay()
obj = upi()
obj.pay()
```
